```python
from TTS.api import TTS

tts = TTS(
    "tts_models/multilingual/multi-dataset/xtts_v2"
)

print("Model loaded")
```

```python
from pathlib import Path
from collections import defaultdict

speaker_files = defaultdict(list)

for wav in wav_files:
    speaker_id = wav.parent.name
    speaker_files[speaker_id].append(wav)

for speaker, files in list(speaker_files.items())[:5]:
    print(
        speaker,
        len(files),
        files[:3]
    )
```

```python
speaker_id = "MGMM0"

reference_audio = speaker_files[speaker_id][0]

print("Speaker:", speaker_id)
print("Reference:", reference_audio)
```

```python
from IPython.display import Audio, display

display(
    Audio( 
        filename=str(reference_audio)
    )
)
```

```python
text = "The quick brown fox jumps over the lazy dog."

output_file = "cloned_MGMM0.WAV.wav"

tts.tts_to_file(
    text=text,
    speaker_wav=str(reference_audio),
    language="en",
    file_path=output_file
)

print("Generated:", output_file)
```
